### Crear el modelo de regresión logística

```python
from sklearn.linear_model import LogisticRegression
```

### Modelo $$ln({P}/{1-P})=\beta_0+\beta_1(BloodPressure)$$

```python
X = df['BloodPressure'].values.reshape(-1, 1)
y = df['HeartDisease']

logit = LogisticRegression()

logit.fit(X, y)

print("Beta cero:", logit.intercept_[0])
print("Beta uno:", logit.coef_[0][0])
```

### Modelo con todas las variables

```python
X1 = df.drop(columns=['HeartDisease'])
y1 = df['HeartDisease']

logit_total = LogisticRegression()
    
logit_total.fit(X1, y1)

print("Beta cero:", logit_total.intercept_[0])
print("Betas uno a n:", logit_total.coef_[0])
```

### Calcular los accuracy, precision, recall y specificity para diferentes threshold (0.3, 0.5, 0.8)

```python
threshold = 0.3
ypred = (logit_total.predict_proba(X1)[:, 1] >= threshold).astype(int)
def classify_row(row):
    if row["ypred"] == 1 and row["y"] == 1:
        return "TP"
    elif row["ypred"] == 1 and row["y"] == 0:
        return "FP"
    elif row["ypred"] == 0 and row["y"] == 0:
        return "TN"
    elif row["ypred"] == 0 and row["y"] == 1:
        return "FN"

table = pd.DataFrame({
    "predict_proba": logit_total.predict_proba(X1)[:, 1],
    "ypred": ypred,
    "y": y
})
table["classification"] = table.apply(classify_row, axis=1)
print(table)

#Calcular Accuracy, Recall, Precision y Specificity
TP = np.sum((table["ypred"] == 1) & (table["y"] == 1))
FP = np.sum((table["ypred"] == 1) & (table["y"] == 0))
TN = np.sum((table["ypred"] == 0) & (table["y"] == 0))
FN = np.sum((table["ypred"] == 0) & (table["y"] == 1))

accuracy = (TP + TN) / (TP + FP + TN + FN)
recall = TP / (TP + FN) if (TP + FN) != 0 else 0
precision = TP / (TP + FP) if (TP + FP) != 0 else 0
specificity = TN / (TN + FP) if (TN + FP) != 0 else 0

print(f"Accuracy: {accuracy}")
print(f"Recall: {recall}")
print(f"Precision: {precision}")
print(f"Specificity: {specificity}")
```
