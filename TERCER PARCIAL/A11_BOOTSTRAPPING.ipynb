### Ejemplo de resample

```python
X = df[["TV", "radio", "newspaper"]]

Xb, yb = resample(X, y, random_state=5)
print(Xb.head())
print(yb.head())
```

### Regresión múltiple simple con bootstrapping

```python
from sklearn.linear_model import LinearRegression
```

```python

model = LinearRegression()
model.fit(Xb, yb)
print("Coeficientes:", model.coef_)
print("Intercepto:", model.intercept_)
```

### Repitiendo lo mismo con otra semilla

```python
Xb, yb = resample(X, y, random_state=6)
print(Xb.head())
print(yb.head())

model = LinearRegression()
model.fit(Xb, yb)
print("Coeficientes:", model.coef_)
print("Intercepto:", model.intercept_)
```

```python
Xb, yb = resample(X, y, random_state=7)
print(Xb.head())
print(yb.head())

model = LinearRegression()
model.fit(Xb, yb)
print("Coeficientes:", model.coef_)
print("Intercepto:", model.intercept_)
```

### Iterando 10,000 veces

```python
results = []
intercepto = []
betas_tv = []
betas_radio = []
betas_np = []
for i in range(10000):
    Xb, yb = resample(X, y, random_state=i)
    model = LinearRegression()
    model.fit(Xb, yb)
    intercepto.append(model.intercept_)
    betas_tv.append(model.coef_[0])
    betas_radio.append(model.coef_[1])
    betas_np.append(model.coef_[2])
    results.append({
        "Intercepto": model.intercept_,
        "Beta TV": model.coef_[0],
        "Beta radio": model.coef_[1],
        "Beta np": model.coef_[2]
    })
results_df = pd.DataFrame(results)
results_df.insert(0, "Iteración", range(len(results_df)))
results_df.head()
```

```python
np.mean(intercepto), np.mean(betas_tv), np.mean(betas_radio), np.mean(betas_np)
```
