```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression

df1 = pd.read_csv('/Users/claudioalejoencarnacionmartinez/Documents/SEMESTRE OTOÑO 2026/LAB APRENDIZAJE ESTADÍSTICO/bateria_tableta.txt')
df1.head()
```

```python
plt.scatter(df1['carga'], df1['duracion'])
plt.ylabel('Duración')
plt.xlabel('Carga')
```

```python
Xv1 = np.reshape(df1['carga'].values, [-1, 1])
y1 = df1['duracion'].values  
Xv1
```

GRADO 1

```python
X1 = Xv1

lr = LinearRegression()
lr.fit(X1, y1)

print("Intercepto (grado 1):", lr.intercept_)
print("Coeficientes (grado 1):", lr.coef_)

plt.scatter(Xv1, y1, label='Datos originales')
plt.scatter(df1['carga'], lr.predict(X1), color='red')
plt.xlabel('Carga')
plt.ylabel('Duración')
plt.title('Regresión polinomial grado 1')
plt.show()
```

GRADO 5

```python
X5 = np.hstack((Xv1, Xv1**2, Xv1**3, Xv1**4, Xv1**5))
X5
```

```python
lr = LinearRegression()
lr.fit(X5, y1)

print("Intercepto (grado 5):", lr.intercept_)
print("Coeficientes (grado 5):", lr.coef_)

plt.scatter(Xv1, y1, label='Datos originales')
plt.scatter(df1['carga'], lr.predict(X5), color='red')
plt.xlabel('Carga')
plt.ylabel('Duración')
plt.title('Regresión polinomial grado 5')
plt.show()
```

GRADO 9

```python
X9 = np.hstack((Xv1, Xv1**2, Xv1**3, Xv1**4, Xv1**5, Xv1**6, Xv1**7, Xv1**8, Xv1**9))
X9
```

```python
lr = LinearRegression()
lr.fit(X9, y1)

print("Intercepto (grado 9):", lr.intercept_)
print("Coeficientes (grado 9):", lr.coef_)

plt.scatter(Xv1, y1, label='Datos originales')
plt.scatter(df1['carga'], lr.predict(X9), color='red')
plt.xlabel('Carga')
plt.ylabel('Duración')
plt.title('Regresión polinomial grado 9')
plt.show()
```
