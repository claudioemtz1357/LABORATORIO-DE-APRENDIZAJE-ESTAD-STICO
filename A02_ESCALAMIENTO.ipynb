```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression

dfbateria = pd.read_csv('/Users/claudioalejoencarnacionmartinez/Documents/SEMESTRE OTOÑO 2026/LAB APRENDIZAJE ESTADÍSTICO/LABORATORIO-DE-APRENDIZAJE-ESTAD-STICO/bateria_tableta.txt')
dfbateria.head()
```

```python
dfbateria.info()
```

```python
dfbateria.describe()
```

```python
dfbateria_mod = dfbateria.copy()
```

```python
dfbateria_mod['carga'] = (dfbateria_mod['carga'] / 11.72) + 1
dfbateria_mod.head()
```

```python
dfbateria_mod.describe()
```

Regresión lineal grado 9
---

```python
#Regresion lineal grado 9
Xv_bateria = np.reshape(dfbateria_mod['carga'].values, [-1, 1])
y_bateria = dfbateria_mod['duracion'].values  

X9_bateria = np.hstack((Xv_bateria, Xv_bateria**2, Xv_bateria**3, Xv_bateria**4, Xv_bateria**5, Xv_bateria**6, Xv_bateria**7, Xv_bateria**8, Xv_bateria**9))
X9_bateria
```

```python
lr_bateria = LinearRegression()
lr_bateria.fit(X9_bateria, y_bateria)

print("Intercepto (grado 9):", lr_bateria.intercept_)
print("Coeficientes (grado 9):", lr_bateria.coef_)

plt.scatter((Xv_bateria-1)*11.72, y_bateria, label='Datos originales')
plt.scatter((dfbateria_mod['carga']-1)*11.72, lr_bateria.predict(X9_bateria), color='red')
plt.xlabel('Carga')
plt.ylabel('Duración')
plt.title('Regresión polinomial grado 9')
plt.show()
```

Regresión grado 9 (Salarios)
---

```python
df_salariosmod = df.copy()
df_salariosmod.head()
```

```python
df_salariosmod['Periodo'] = (df_salariosmod['Periodo'] / 2018) + 1
df_salariosmod.head()
```

```python
df_salariosmod.describe()
```

```python
#Regresión grado 9
Xv_salarios = np.reshape(df_salariosmod['Periodo'].values, [-1, 1])
y_salarios = df_salariosmod['Salario'].values

X9_salarios = np.hstack((Xv_salarios, Xv_salarios**2, Xv_salarios**3, Xv_salarios**4, Xv_salarios**5, Xv_salarios**6, Xv_salarios**7, Xv_salarios**8, Xv_salarios**9))
X9_salarios
```

```python
lr_salarios = LinearRegression()
lr_salarios.fit(X9_salarios, y_salarios)

print("Intercepto (grado 9):", lr_salarios.intercept_)
print("Coeficientes (grado 9):", lr_salarios.coef_)

plt.scatter((Xv_salarios-1)*2018, y_salarios, label='Datos originales')
plt.plot((df_salariosmod['Periodo']-1)*2018, lr_salarios.predict(X9_salarios), color='red')
plt.xlabel('Periodo')
plt.ylabel('Salario')
plt.title('Regresión polinomial grado 9 (Salarios, escalado)')
plt.show()
```

ESTÁNDAR
---

Hacer que la media sea 0, y la desviación estándar sea 1

```python
df_bat_estandar = dfbateria.copy()
df_bat_estandar.head()
```

```python
Xbarra_bat = dfbateria['carga'].mean()
std = dfbateria['carga'].std()
#Dividir cada dato de carga entre n y restar el promedio, para que al final el promedio sea 0
df_bat_estandar['carga'] -= Xbarra_bat
df_bat_estandar['carga'] /= std
df_bat_estandar.head()
```

```python
print("Media de la carga:", df_bat_estandar['carga'].mean(), "Desviación estándar de la carga:", df_bat_estandar['carga'].std())
```

ESCALAMIENTO CON SKLEARN
---

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
```

```python
df_salarios_sklearn = df.copy()
df_salarios_sklearn.head()
```

```python
scaler = StandardScaler()
X_scaler = df_salarios_sklearn['Periodo'].values.reshape(-1, 1)
scaler.fit(X_scaler)
X_scaled = scaler.transform(X_scaler)
```

```python

X9_salarios_scaled = np.hstack((X_scaled, X_scaled**2, X_scaled**3, X_scaled**4, X_scaled**5, X_scaled**6, X_scaled**7, X_scaled**8, X_scaled**9))

lr_salarios_scaled = LinearRegression()
lr_salarios_scaled.fit(X9_salarios_scaled, df_salarios_sklearn['Salario'].values)

print("Intercepto (grado 9, escalado):", lr_salarios_scaled.intercept_)
print("Coeficientes (grado 9, escalado):", lr_salarios_scaled.coef_)

# Plot the results
plt.scatter(df_salarios_sklearn['Periodo'], df_salarios_sklearn['Salario'], label='Datos originales', color='blue')
plt.plot(df_salarios_sklearn['Periodo'], lr_salarios_scaled.predict(X9_salarios_scaled), color='red', linewidth=2, label='Regresión polinomial grado 9')
plt.xlabel('Periodo')
plt.ylabel('Salario')
plt.title('Regresión polinomial grado 9 (Salarios, con StandardScaler)')
plt.legend()
plt.show()
```
