### Con librerías básicas

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

df['SABOR'].unique() me da cuantos datos diferentes hay en la columna

```python
print(df['SABOR'].unique())
print(df['TAMANO'].unique())
print(df['HIELO'].unique())
print(df['CALIFICACION'].unique())
```

Cambiar variables de tamaño y hielo a valores 0 y 1

```python
#Codificar la columna TAMANO, para que CH sea 0 y G sea 1
df['TAMANO'] = df['TAMANO'].map({'CH': 0, 'G': 1})
df['TAMANO'].unique()
```

```python
#Codificar la columna HIELO, para que SI sea 1 y NO sea 0
df['HIELO'] = df['HIELO'].map({'SI': 1, 'NO': 0})
df['HIELO'].unique()
```

```python
df.head()
```

```python
#Separar la columna SABOR en 5 columnas, una para cada sabor, empezar con columna NARANJA, tomando si el dato contiene "N" o no
df['NARANJA'] = df['SABOR'].str.contains('N')
df['COCO'] = df['SABOR'].str.contains('C')
df['FRESAPINA'] = df['SABOR'].str.contains('FP')
df['KAHLUA'] = df['SABOR'].str.contains('K')
df['SANDIA'] = df['SABOR'].str.contains('S')
df.head()
```

```python
#Eliminar columna SABOR
df = df.drop(columns=['SABOR'])
df.head()
```

```python
#Reordenar columnas al principio, primero las columnas de sabor, luego TAMANO, HIELO y CALIFICACION
df = df[['NARANJA', 'COCO', 'KAHLUA', 'FRESAPINA', 'SANDIA', 'TAMANO', 'HIELO', 'CALIFICACION']]
df.head()
```

## Realizar la regresión con la librería con OLS

```python
import statsmodels.api as sm
```

```python
#Regresión lineal con OLS, usando todas las columnas como variables independientes y CALIFICACION como variable dependiente
X = df[['NARANJA', 'COCO', 'KAHLUA', 'FRESAPINA', 'SANDIA', 'TAMANO', 'HIELO']]
X = sm.add_constant(X)
y = df['CALIFICACION']
model = sm.OLS(y, X).fit()
model.summary()
```

## Realizar la codificación con sklearn

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
df_sk = pd.read_csv('/Users/claudioalejoencarnacionmartinez/Documents/SEMESTRE OTOÑO 2026/LAB APRENDIZAJE ESTADÍSTICO/LABORATORIO-DE-APRENDIZAJE-ESTAD-STICO/FEO.csv')
df_sk.head()
```

```python
#Codificar la columna TAMANO, para que CH sea 0 y G sea 1 con OrdinalEncoder
ordinal_encoder = OrdinalEncoder(categories=[['CH', 'G']])
df_sk['TAMANO'] = ordinal_encoder.fit_transform(df_sk[['TAMANO']])
df_sk['TAMANO'].unique()
```

```python
#Codificar la columna HIELO, para que SI sea 1 y NO sea 0 con OrdinalEncoder
ordinal_encoderHIELO = OrdinalEncoder(categories=[['NO', 'SI']])
df_sk['HIELO'] = ordinal_encoderHIELO.fit_transform(df_sk[['HIELO']])
df_sk['HIELO'].unique()
```

```python
df_sk.head(10)
```

```python
#Codificar la columna SABOR con OneHotEncoder con 5 columnas, una para cada sabor, empezando con NARANJA, COCO, KAHLUA, FRESAPINA y SANDIA
onehot_encoder = OneHotEncoder(categories=[['N', 'C', 'K', 'FP', 'S']])
df_sk[['NARANJA', 'COCO', 'KAHLUA', 'FRESAPINA', 'SANDIA']] = onehot_encoder.fit_transform(df_sk[['SABOR']]).toarray()
df_sk.head(10)
```

```python
#Borrar columna de SABOR
df_sk = df_sk.drop(columns=['SABOR'])
df_sk.head(10)
```

```python
#Reordenar columnas al principio, primero las columnas de sabor, luego TAMANO, HIELO y CALIFICACION
df_sk = df_sk[['NARANJA', 'COCO', 'KAHLUA', 'FRESAPINA', 'SANDIA', 'TAMANO', 'HIELO', 'CALIFICACION']]
df_sk.head(5)
```

Rehaciendo la regresión

```python
X_sk = df_sk[['NARANJA', 'COCO', 'KAHLUA', 'FRESAPINA', 'SANDIA', 'TAMANO', 'HIELO']]
X_sk = sm.add_constant(X_sk)
y_sk = df_sk['CALIFICACION']
model_sk = sm.OLS(y_sk, X_sk).fit()
model_sk.summary()
```
